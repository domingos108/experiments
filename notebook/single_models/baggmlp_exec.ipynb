```python
import warnings
from statsmodels.tools.sm_exceptions import InterpolationWarning
warnings.simplefilter('ignore', InterpolationWarning)

from sklearn.neural_network import MLPRegressor
from sklearn.ensemble import BaggingRegressor

from model import single_ml_model_exp, grid_search_exp

import config
%load_ext autoreload
%autoreload 2

for horizon in [24, 48, 72]:
    for n_estimators in [10, 30, 50]:
        for max_samples in [0.5, 0.8]:
            for max_features in [0.5, 0.8]:
                model = BaggingRegressor(n_jobs=-1)
                
                experiment_id = 'hourwindspeed'
                model_name = f'mlpbagg{n_estimators}{max_samples}{max_features}'
                normalize = True
                force = False
                model_exec = 10
                
                experiment_params = {
                    'diff_kpss': False,
                    'horizon': horizon
                }
                
                model_parameters = {
                            'estimator': [MLPRegressor(activation='logistic', solver='lbfgs', max_iter = 1000, hidden_layer_sizes=10) ,
                                          MLPRegressor(activation='logistic', solver='lbfgs', max_iter = 1000, hidden_layer_sizes=20) ,
                                          MLPRegressor(activation='logistic', solver='lbfgs', max_iter = 1000, hidden_layer_sizes=50) 
                                         ], 
                            'n_estimators': [n_estimators],
                            'max_features': [max_features],
                            'max_samples': [max_samples]
                            
                        }
        
                grid_search_exp.grid_seach_multiple_bases(
                     single_ml_model_exp.SKlearnModel, model, normalize, model_parameters,experiment_params,
                     model_exec, model_name, experiment_id, force = force
                )
```
