```python
import os
os.environ['PYTHONWARNINGS'] = 'ignore'

import warnings
from statsmodels.tools.sm_exceptions import InterpolationWarning
from sklearn.exceptions import ConvergenceWarning

warnings.simplefilter('ignore', InterpolationWarning)
warnings.filterwarnings("ignore", category=ConvergenceWarning)

from sklearn.linear_model import PoissonRegressor
from sklearn.ensemble import BaggingRegressor

from model import single_ml_model_exp, grid_search_exp

import config
%load_ext autoreload
%autoreload 2


for n_estimators in [50]:
    for max_samples in [0.8]:
        for max_features in [0.8]:
            model = BaggingRegressor(n_jobs=-1)
            
            experiment_id = 'hourwindspeed'
            model_name = f'poissonbagg{n_estimators}{max_samples}{max_features}'
            normalize = True
            force = True
            model_exec = 10
            
            experiment_params = {
                'diff_kpss': False,
                'horizon': 24
            }
            
            model_parameters = {
                        'estimator': [
                             PoissonRegressor(max_iter=100, alpha=0.1),
                            PoissonRegressor(max_iter=500, alpha=0.1),
                            PoissonRegressor(max_iter=1000, alpha=0.1),
                            PoissonRegressor(max_iter=100, alpha=0.4),
                            PoissonRegressor(max_iter=500, alpha=0.4),
                            PoissonRegressor(max_iter=1000, alpha=0.4),
                            PoissonRegressor(max_iter=100, alpha=0.8),
                            PoissonRegressor(max_iter=500, alpha=0.8),
                            PoissonRegressor(max_iter=1000, alpha=0.8)
                        ], 
                        'n_estimators': [n_estimators],
                        'max_features': [max_features],
                        'max_samples': [max_samples]
                        
                    }
    
            grid_search_exp.grid_seach_multiple_bases(
                 single_ml_model_exp.SKlearnModel, model, normalize, model_parameters,experiment_params,
                 model_exec, model_name, experiment_id, force = force
            )
```
